## Model


### Governing equations

**Shallow water equations.** Conservative form of the equations

$$\left\{ \begin{aligned}
  & \partial_t \zeta + \nabla \cdot ( \zeta \mathbf{u} ) = 0 \\
  & \partial_t ( \zeta \mathbf{u} ) + \nabla \cdot \left( \zeta \mathbf{u} \otimes \mathbf{u} + \frac{g \zeta^2}{2} \mathbb{I} \right) = \text{variable depth} + \text{non inertial + tidal + dissipation} \ ,
\end{aligned}\right.$$

while convective form reads

$$\left\{ \begin{aligned}
  & \partial_t \zeta + \nabla \cdot ( \zeta \mathbf{u} ) = 0 \\
  & \zeta \partial_t  \mathbf{u} + \zeta \mathbf{u} \cdot \nabla \mathbf{u} + g \zeta \nabla \zeta = \text{variable depth} + \text{non inertial + tidal + dissipation}
\end{aligned}\right.$$

or, dividing the momentum equation by $\zeta$

$$\left\{ \begin{aligned}
  & \partial_t \zeta + \nabla \cdot ( \zeta \mathbf{u} ) = 0 \\
  & \partial_t  \mathbf{u} + \mathbf{u} \cdot \nabla \mathbf{u} + g \nabla \zeta = \text{variable depth} + \text{non inertial + tidal + dissipation}
\end{aligned}\right.$$

The sum of Moon gravitational pull and non-inertial force per unit-mass, with constant angular velocity of the non-inertial reference frame w.r.t. the inertial ref.frame $\boldsymbol\omega_{1/0}$ s.t. $\boldsymbol\alpha_{1/0} = \mathbf{0}$, reads

$$-\mathbf{a}^{(0)}_{P/O} - 2 \boldsymbol\omega_{1/0} \times \mathbf{v}^{(1)} - \boldsymbol\omega_{1/0} \times \boldsymbol\omega_{1/0} \times (P-Q) \ .$$

Using spherical coordinates $\boldsymbol\omega_{1/0} = \Omega_{1/0} \hat{\mathbf{z}}$, $\mathbf{v}^{(1)} = v_{\varphi}\hat{\boldsymbol\varphi} + v_{\theta}\hat{\boldsymbol\theta}$, $\mathbf{P-Q} = a \hat{\mathbf{R}}$,

* centrifugal acceleration

   $$\boldsymbol\omega \times \boldsymbol\omega \times (P-Q) = \Omega^2 a \, \hat{\mathbf{z}} \sin \varphi \times \hat{\boldsymbol\theta} = - \Omega^2 a \sin \varphi \hat{\boldsymbol\rho} = \Omega^2 a \left( - \sin^2 \varphi \, \hat{\mathbf{R}} - \sin \varphi \cos \varphi \, \hat{\boldsymbol\varphi} \right) $$

* Coriolis force (divided by $2$)

   $$\boldsymbol\omega_{1/0} \times \mathbf{v}^{(1)} = \Omega \hat{\mathbf{z}} \times ( v_{\varphi}\hat{\boldsymbol\varphi} + v_{\theta}\hat{\boldsymbol\theta}) = \Omega v_\varphi \cos \varphi \hat{\boldsymbol\theta} - \Omega v_\theta \hat{\boldsymbol\rho} =  \Omega v_\varphi \cos \varphi \hat{\boldsymbol\theta} - \Omega v_\theta \left( \sin \varphi \, \hat{\mathbf{R}} + \cos \varphi \, \hat{\boldsymbol\varphi} \right) \ ,$$

  i.e., using lat $\phi = \frac{\pi}{2} - \varphi$,

  $$2 \boldsymbol\omega_{1/0} \times \mathbf{v}^{(1)} = - 2 \Omega v_\theta \cos \phi \, \hat{\mathbf{R}} + 2 \Omega \, \sin \phi \left( -v_\theta \hat{\boldsymbol\varphi} + v_\varphi \hat{\boldsymbol\theta} \right) $$

* linear dissipation $- r \mathbf{u}$


**Steady reference state.** ... centrifugal force is constant in the non-inertial reference frame, so it only acts on the steady reference state and not in the unsteady linearized equations around that.

**Linearized equations** around the reference state $\left.\left(\zeta, \mathbf{u}\right)\right|_{\rm ref} = (\zeta_0(\mathbf{r}), \mathbf{0})$

$$\left\{ \begin{aligned}
  & \partial_t \zeta' + \nabla \cdot \left( \zeta_0  \mathbf{u}'  \right) = 0 \\
  & \partial_t \mathbf{u}' + g \nabla \zeta' = -2 \boldsymbol\omega_{1/0} \times \mathbf{u}' - r \mathbf{u}' + \text{tidal forces}
\end{aligned}\right.$$

+ i.c + b.c.

**Frequency domain.** Frequency response of the system is studied in frequency domain, with Fourier transform of the functions, $\hat{f}(\mathbf{r}, \omega) := \mathcal{F}\{ f(\mathbf{r}, t) \}(\omega)$, and the linearized equations for $\left( \hat{\zeta}, \hat{\mathbf{u}} \right)(\mathbf{r}, \omega)$,

$$\left\{ \begin{aligned}
  & i \omega \hat{\zeta} + \nabla \cdot \left( \zeta_0  \hat{\mathbf{u}} \right) = 0 \\
  & i \omega \hat{\mathbf{u}} + g \nabla \hat{\zeta} + 2 \boldsymbol\omega_{1/0} \times \hat{\mathbf{u}} + r \hat{\mathbf{u}} = \hat{\mathbf{g}}_{\rm tidal}(\omega) \ .
\end{aligned}\right.$$

+ i.c + b.c.

#### Characteristics of the hyperbolic problem and flux boundary conditions (e.g. for Finite Volume Methods)

**n-dimensional linearized shallow water equations.**

$$\left\{ \begin{aligned}
  & \partial_t \zeta' + \nabla \cdot ( \zeta_0 \mathbf{u}' ) = 0 \\
  & \partial_t ( \zeta_0 \mathbf{u}' ) + \nabla \cdot \left( g \zeta_0 \zeta' \mathbb{I} \right) = \text{variable depth} + \text{non inertial + tidal + dissipation} \ ,
\end{aligned}\right.$$


**1-dimensional linearized shallow water equations.**

$$\left\{ \begin{aligned}
  & \partial_t \zeta' + \partial_x ( \zeta_0 u' ) = 0 \\
  & \partial_t ( \zeta_0 u' ) + \partial_x \left( g \zeta_0 \zeta' \right) = \text{variable depth} + \text{non inertial + tidal + dissipation} \ ,
\end{aligned}\right.$$



*Conservative form*

$$
\partial_t \begin{bmatrix} \zeta' \\ \zeta_0 u' \end{bmatrix} + \partial_x \begin{bmatrix} \zeta_0 u' \\ g \zeta_0 \zeta' \end{bmatrix} = \begin{bmatrix} 0 \\ \text{non inertial + tidal + dissipation} \end{bmatrix}
$$

*Convective form*

$$
\partial_t \begin{bmatrix} \zeta' \\ \zeta_0 u' \end{bmatrix} + \begin{bmatrix} 0 & 1 \\ g \zeta_0 & 0 \end{bmatrix} \partial_x \begin{bmatrix} \zeta' \\ \zeta_0 u' \end{bmatrix} = - \begin{bmatrix} 0 \\ \partial_x( g \zeta_0 ) \, \zeta' \end{bmatrix} + \begin{bmatrix} 0 \\ \text{non inertial + tidal + dissipation} \end{bmatrix}
$$

The characteristic velocities of the linearized hyperbolic problem are the eigenvalues of the advection matrix $\mathbf{A}$, here

$$0 = \left| \mathbf{A} - s_{1,2} \mathbf{I} \right| = s^2 - g\zeta_0 \quad \rightarrow \quad s_{1,2} = \mp \sqrt{g \zeta_0} =: \mp c_0 \ ,$$

with corresponding right and left eigenvectors collected as the columns of matrices $\mathbf{R}$ and $\mathbf{L}$,

$$\begin{aligned}
  \mathbf{R} = \begin{bmatrix} \zeta_0 & \zeta_0 \\ \zeta_0 c_0 & - \zeta_0 c_0 \end{bmatrix}
  \quad , \quad 
  \mathbf{L} = \mathbf{R}^{-1} = \frac{1}{2 \zeta_0 c_0} \begin{bmatrix} -c_0 & -1 \\ - c_0 & 1 \end{bmatrix}
\end{aligned}$$


Since the 2 characteristic velocities have opposite sign, in every point of the domain there's a family of characteristics moving to the right and one family moving to the left. At the boundaries of the domain, there's always a familiy of characteristics leaving the domain, and one family entering the domain.

This observation is used in the implementation of boundary conditions for finite volume methods.


**Solid boundaries.** In order to implement zero-mass flux boundary condition through a boundary at rest, $\mathbf{u} \cdot \hat{\mathbf{n}} = 0$, a *ghost cell* with state $\mathbf{u}_{g} = \left( \zeta_\partial, -\zeta_\partial u_\partial \right)$ can be defined as a function of the state $\mathbf{u}_\partial$ of the cell at the boundary, so that Roe numerical flux reads

$$\mathbf{F}^{\rm Roe} = \frac{1}{2} \left\{ \mathbf{F}_g + \mathbf{F}_\partial - \left| \hat{\mathbf{A}} \right| ( \mathbf{u}_g - \mathbf{u}_\partial ) \right\} = \begin{bmatrix} 0 \\ \dots \end{bmatrix} \ .$$


**Non-reflective boundary conditions (Flather?).**

$$\partial_t \mathbf{u} + \mathbf{A} \partial_x \mathbf{u} = \mathbf{F}(\mathbf{u})$$

Diagonalization of $\mathbf{A} = \mathbf{R} \boldsymbol\Lambda \mathbf{L}$, and definition of characteristic variables $\mathbf{v} = \mathbf{L} \mathbf{u}$, governed by a the equations

$$\partial_t \mathbf{v} + \boldsymbol\Lambda \partial_x \mathbf{v} = \mathbf{F}(\mathbf{R} \mathbf{v})$$

### Tidal force

* Semi-diurnal ($\approx 12$ hours)
  * Principal lunar ($\rm M_2$, period $T_{\rm M_2} \simeq 12.42 \, {\rm h}$)
  * Principal solar ($\rm S_2$, period $T_{\rm S_2} \simeq 12.00 \, {\rm h}$)
  * Larger lunar elliptic ($\rm N_2$), driven by changes in Earth-Moon distance along their elliptical orbit
* Diurnal ($\approx 24$ hours)
  * Luni-solar diurnal ($\rm K_1$, period $T_{\rm K_1} \simeq 23 \, {\rm h} \, 56'$, sidereal day, due to declination changes of the Moon and Sun)
  * Principal lunar ($\rm O_1$, period $T_{\rm M_2} \simeq 25 \, {\rm h} \, 49'$)
  * Principal solar ($\rm P_1$, period $T_{\rm M_2} \simeq 14 \, {\rm h} \, 4'$)
* Long-period: forth-nighlty, monthly, semi-annual, annual, multi-years
